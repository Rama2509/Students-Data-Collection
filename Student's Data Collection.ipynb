```python
# Latihan 1
import pandas as pd
print('='*150)
print(f'{'='*54} Selamat Datang di Laman Data Base Siswa {'='*55} \n')
jumlah_data = int(input("Berapa banyak siswa yang akan Anda masukkan datanya? (dalam angka) : "))

daftar_nama = []
daftar_tahun_masuk = []
daftar_kelas = []
daftar_lahir = []
daftar_kota = []
daftar_nomor = []
daftar_email = []
daftar_wali = []
daftar_nomor_wali = []

for i in range(jumlah_data):
    print(f'\nData ke-{i+1}')
    nama = input('Masukkan Nama Siswa : ')
    tahun_masuk = input('Masukkan Tahun Masuk Siswa : ')
    kelas = input('Masukkan Kelas Siswa : ')
    lahir = input('Masukkan Tanggal Lahir Siswa (yyyy/mm/dd) : ')
    kota = input('Masukkan Kota Tempat Tinggal : ')
    nomor = input('Masukkan Nomor HP Siswa (08xxx ... ) : ')
    email = str(input('Masukkan Email Siswa : '))
    wali = input('Masukkan Nama Wali Siswa : ')
    nomor_wali = input('Masukkan Nomor Wali Siswa : ')

    daftar_nama.append(nama)
    daftar_tahun_masuk.append(tahun_masuk)
    daftar_kelas.append(kelas)
    daftar_lahir.append(lahir)
    daftar_kota.append (kota)
    daftar_nomor.append(nomor)
    daftar_email.append(email)
    daftar_wali.append(wali)
    daftar_nomor_wali.append(nomor_wali)

data_siswa = {
    'Nama' : daftar_nama,
    'Tahun Masuk' : daftar_tahun_masuk,
    'Kelas' : daftar_kelas,
    'Tanggal Lahir' : daftar_lahir,
    'Kota' : daftar_kota,
    'Nomor HP Siswa' : daftar_nomor,
    'Email Siswa' : daftar_email,
    'Wali' : daftar_wali,
    'Nomor HP Wali' : daftar_nomor_wali
}

data = pd.DataFrame(data_siswa)
data.index = range(1, len(data)+1)
print(f'\nData Base Siswa')
display(data)
    
```
